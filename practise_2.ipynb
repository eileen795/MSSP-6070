```python
letters = ['a',  'b',  'c', 'd']
print(letters[int(int('3'*2)//11)]) #int'3'*2=33, int(33//11) floor division of 33/11 is 3, so it means finding the index 3 in the list
print( letters[-1]) #this means shows the second index counted backwards
print(letters[:2]) #this means shows the element till the third index, and the third one is not included
```

```python
def comma_code(collection):
  if len(collection)==0:
    return ''
  elif len(collection)==1:
    return str(collection[0])
  return f"{', '.join(str(item) for item in collection[:-1])}, and {str(collection[-1])}"

spam = ['apples', 'bananas', 'tofu', 'cats']
print(comma_code(collection=spam))
```
