```python

import pandas as pd
from google.colab import drive

# Step 1: Mount Google Drive
drive.mount('/content/drive')

# Step 2: Read the dataset
file_path = '/content/drive/MyDrive/MSSP607/Modules/week6/Student Performance Data.csv'

df = pd.read_csv(file_path)

# Step 3: Display original data
print("Original DataFrame:")
print(df.head())

# Step 4: Define a lambda function
student_performance = lambda data: data.assign(
    average_score=data[
        ['math score', 'reading score', 'writing score']
    ].mean(axis=1)
)

# Step 5: Create a new DataFrame
new_df = student_performance(df)

# Step 6: Display the new DataFrame
print("\nNew DataFrame:")
print(new_df.head())
```
