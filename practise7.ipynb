```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Load data directly from GitHub
url = "https://raw.githubusercontent.com/eileen795/MSSP-6070/main/Modules/week7/InternetSales.csv"
df = pd.read_csv(url, encoding="cp1252")
```

```python
df.head()
```

```python
# Step 2: Convert OrderDate to datetime
df['OrderDate'] = pd.to_datetime(df['OrderDate'])

# Step 3: Extract year from OrderDate
df['Year'] = df['OrderDate'].dt.year

# Step 4: Calculate yearly total sales
yearly_sales = df.groupby('Year')['SalesAmount'].sum()
# Step 5: Create a bar chart
plt.figure(figsize=(9, 6))
plt.bar(yearly_sales.index, yearly_sales.values)

plt.title('Yearly Internet Sales')
plt.xlabel('Year')
plt.ylabel('Total Sales Amount ($)')
plt.xticks(yearly_sales.index)
plt.ticklabel_format(style='plain', axis='y')
plt.grid(axis='y', alpha=0.3)

# Step 6: Save the active figure as PDF
plt.savefig('Yearly_Internet_Sales.pdf',
            format='pdf', bbox_inches='tight')

# Step 7: Display the chart
plt.show()
```
