```python
from google.colab import drive
drive.mount('/content/drive')
import glob

# find parsed csv path
import glob
path = glob.glob('/content/drive/MyDrive/**/parsed.csv', recursive=True)
print(path)
```

```python
import pandas as pd

df = pd.read_csv(path[0])   # path[0] 就是上一步找到的路径
df.head()
```

```python
# a. Find the 95th percentile of earthquake magnitude in Japan using magType of 'mb'
japan_mb = df.query("parsed_place == 'Japan' and magType == 'mb'")
p95 = japan_mb['mag'].quantile(0.95)
print(p95)
```

```python
#b. Find the percentage of earthquakes in Indonesia that were coupled with tsunamis.
indonesia = df.query("parsed_place == 'Indonesia'")
pct = indonesia['tsunami'].mean() * 100
print(f"{pct:.2f}%")
```

```python
#c. Get summary statistics for earthquakes in Nevada
nevada_summary = df.query("parsed_place == 'Nevada'").describe()
print(nevada_summary)
```

```python
#d add a colom
ring_of_fire = [
    'Bolivia', 'Chile', 'Ecuador', 'Peru', 'Costa Rica', 'Guatemala',
    '(?<!New )Mexico',
    'Japan', 'Philippines', 'Indonesia', 'New Zealand',
    'Antarctic',
    'Canada', 'Fiji', 'Alaska', 'Washington', 'California',
    'Russia', 'Taiwan', 'Tonga', 'Kermadec Islands'
]

pattern = '|'.join(ring_of_fire)
df['ring_of_fire'] = df['parsed_place'].str.contains(pattern, na=False)

df[['parsed_place', 'ring_of_fire']].head()
```

```python
#e. Calculate the number of earthquakes in the Ring of Fire locations and the number outside of them,
counts = df['ring_of_fire'].value_counts()
print(counts)
```

```python
# f. Find the tsunami count along the Ring of Fire.
tsunami_rof = df.query('ring_of_fire')['tsunami'].sum()
print(tsunami_rof)
```
