```python
# 1. Import libraries and mount Google Drive
import pandas as pd
from google.colab import drive

drive.mount('/content/drive')


# 2. Set the folder path
path = '/content/drive/MyDrive/MSSP607/Modules/week5/'


# b. Read each file into a dataframe
aapl = pd.read_csv(path + 'aapl.csv')
amzn = pd.read_csv(path + 'amzn.csv')
fb   = pd.read_csv(path + 'fb.csv')
goog = pd.read_csv(path + 'goog.csv')
nflx = pd.read_csv(path + 'nflx.csv')


# c. Add ticker column to each dataframe
aapl['ticker'] = 'AAPL'
amzn['ticker'] = 'AMZN'
fb['ticker']   = 'FB'
goog['ticker'] = 'GOOG'
nflx['ticker'] = 'NFLX'


# d. Append all files into one dataframe
faang = pd.concat(
    [aapl, amzn, fb, goog, nflx],
    ignore_index=True
)


# e. Save the combined dataframe as faang.csv
faang.to_csv(path + 'faang.csv', index=False)


# f. Read faang.csv
faang = pd.read_csv(path + 'faang.csv')

# Convert date column to datetime
faang['date'] = pd.to_datetime(faang['date'])

# Convert volume column to integer
faang['volume'] = faang['volume'].astype(int)

# Sort by date and ticker
faang = faang.sort_values(
    by=['date', 'ticker']
)

faang.head()
```

```python
# g. Find the seven rows with the highest volume
faang.nlargest(7, 'volume')
```

```python
# h. Melt the dataframe
melted = faang.melt(
    id_vars=['date', 'ticker'],
    var_name='variable',
    value_name='value'
)

melted.head()
```
