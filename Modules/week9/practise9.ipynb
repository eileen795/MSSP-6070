```python

# ============================================================
# IMAGE DOWNLOADER
# Search and download photos from Flickr
# ============================================================

# STEP 1: Install required libraries
!pip -q install requests==2.32.4 beautifulsoup4

# STEP 2: Import libraries
import os
import re
import time
import shutil
import requests

from urllib.parse import quote
from bs4 import BeautifulSoup
from google.colab import files

# STEP 3: Set search category
category = "cats"       # Change to dogs, flowers, nature, etc.
max_images = 20         # Maximum number of images to download

# STEP 4: Create download folder
folder = f"{category}_images"
os.makedirs(folder, exist_ok=True)

print("Search category:", category)
print("Download folder:", folder)

# STEP 5: Search Flickr's public photo feed
search_url = (
    "https://www.flickr.com/services/feeds/photos_public.gne"
)

params = {
    "tags": category,
    "tagmode": "all",
    "format": "json",
    "nojsoncallback": 1
}

headers = {
    "User-Agent": "Mozilla/5.0 (Educational Image Downloader)"
}

response = requests.get(
    search_url,
    params=params,
    headers=headers,
    timeout=30
)

response.raise_for_status()

data = response.json()

photos = data.get("items", [])

print(f"Found {len(photos)} photos in the public feed.")

# STEP 6: Download images
count = 0

for photo in photos:

    if count >= max_images:
        break

    image_url = photo.get("media", {}).get("m")

    if not image_url:
        continue

    try:
        image_response = requests.get(
            image_url,
            headers=headers,
            timeout=30
        )

        image_response.raise_for_status()

        content_type = image_response.headers.get(
            "Content-Type", ""
        )

        if not content_type.startswith("image/"):
            continue

        count += 1

        filename = os.path.join(
            folder,
            f"image_{count}.jpg"
        )

        with open(filename, "wb") as file:
            file.write(image_response.content)

        print(f"Downloaded {count}: {filename}")

        # Be polite to the server
        time.sleep(0.5)

    except requests.RequestException as e:
        print("Download failed:", e)

print("\nDownload completed!")
print("Total images downloaded:", count)

# STEP 7: Compress downloaded images into ZIP
if count > 0:

    zip_path = shutil.make_archive(
        f"{category}_images",
        "zip",
        folder
    )

    print("ZIP file created:", zip_path)

    # STEP 8: Download ZIP to your computer
    files.download(zip_path)

else:
    print("No images downloaded. Try another category.")
```
