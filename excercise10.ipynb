```python
import pandas as pd

# Load the student performance dataset
url = "https://raw.githubusercontent.com/eileen795/MSSP-6070/main/Modules/week6/Student%20Performance%20Data.csv"
df = pd.read_csv(url)

# Check the dataset and display the first five rows
print("Dataset shape:", df.shape)
print("\nOriginal DataFrame:")
print(df.head())

# Select the three subjects used to calculate the average score
score_columns = ["math score", "reading score", "writing score"]

# Check that the required columns exist
missing_columns = [col for col in score_columns if col not in df.columns]
if missing_columns:
    raise ValueError(f"Missing required columns: {missing_columns}")

# Add each student's average score without changing the original DataFrame
new_df = df.assign(average_score=df[score_columns].mean(axis=1))

# Display the updated data and summary statistics
print("\nNew DataFrame:")
print(new_df.head())
print("\nAverage score summary:")
print(new_df["average_score"].describe())
```
