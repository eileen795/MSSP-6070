## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
mean = sum(salaries)/len(salaries)
print(mean)
import statistics
print(statistics.mean(salaries))
```

### median

```python
median = sorted(salaries)[len(salaries)//2]
print(median)
```

### mode

```python
mode = max(set(salaries), key=salaries.count)
print(mode)
```

### sample variance
Remember to use Bessel's correction.

```python
variance = sum([(salary - mean)**2 for salary in salaries])/(len(salaries)-1)
print(variance)
```

### sample standard deviation
Remember to use Bessel's correction.

```python
standard_deviation = variance**0.5
print(standard_deviation)
```

## Exercise 6: Calculating more statistics
### range

```python
range = max(salaries) - min(salaries)
print(range)
```

### coefficient of variation

```python
coefficient_of_variation = standard_deviation/mean
print(coefficient_of_variation)
```

### interquartile range

```python
interquartile_range = sorted(salaries)[len(salaries)//4*3] - sorted(salaries)[len(salaries)//4]
print(interquartile_range)
```

### quartile coefficent of dispersion

```python
quartile_coefficient_of_dispersion = interquartile_range/mean
print(quartile_coefficient_of_dispersion)
```

## Exercise 7: Scaling data
### min-max scaling

```python
min_max_scaled = [(salary - min(salaries))/(max(salaries) - min(salaries)) for salary in salaries]
print(min_max_scaled)
```

### standardizing

```python
z_scores = [(salary - mean)/standard_deviation for salary in salaries]
print(z_scores)
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
# covariance between salaries and what?
covariance = sum([(x - mean)*(y - mean) for x, y in zip(salaries, z_scores)])/(len(salaries)-1)
print(covariance)
```

### Pearson correlation coefficient ($\rho$)

```python
pearson_correlation_coefficient = covariance/(standard_deviation*standard_deviation)
print(pearson_correlation_coefficient)
```
